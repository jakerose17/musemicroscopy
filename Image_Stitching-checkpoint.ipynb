# OpenCV: Classical Image Stitching

```python
# Path to the directory containing your images
image_directory = r"C:\Users\Public\musemicroscopy\MUSE_98"

# Load all images in the directory into a list
images = []
for filename in os.listdir(image_directory):
    if filename.endswith('.jpg') or filename.endswith('.png'):  # Add more image formats as needed
        image_path = os.path.join(image_directory, filename)
        img = cv2.imread(image_path)
        if img is not None:
            images.append(img)

# Initialize the stitcher in PANORAMA mode (the default)
stitcher = cv2.Stitcher_create(cv2.Stitcher_PANORAMA)

# Stitch all images
status, stitched_image = stitcher.stitch(images)

# Check the status and display the result
if status == cv2.Stitcher_OK:
    print("Stitching successful.")
    plt.imshow(cv2.cvtColor(stitched_image, cv2.COLOR_BGR2RGB))
    plt.show()
else:
    print("Stitching failed:", status)
```

```python
import clip
import torch
from PIL import Image

# Load CLIP model
model, preprocess = clip.load('ViT-B/32', device='cpu')

# Load the stitched image and the reference image
stitched_img = Image.open("stitched_image.jpg")
reference_img = Image.open("reference_image.jpg")

# Preprocess images
stitched_img_input = preprocess(stitched_img).unsqueeze(0).to(device)
reference_img_input = preprocess(reference_img).unsqueeze(0).to(device)

# Get features from the model
stitched_img_features = model.encode_image(stitched_img_input)
reference_img_features = model.encode_image(reference_img_input)

# Compute similarity between the stitched image and the reference image
similarity = torch.cosine_similarity(stitched_img_features, reference_img_features)
print(f"Similarity between stitched and reference image: {similarity.item() * 100:.2f}%")
```
